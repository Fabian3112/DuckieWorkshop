```python
map = []
directions = []

previous_segment = LaneSegment(
    (0,0), 0 , TileType.Straight)

map.extend(previous_segment.points)
print(f"segment: {previous_segment.map_position}, {np.degrees(previous_segment.orientation)}, {previous_segment.type} start: {previous_segment.points[0]}, end: {previous_segment.points[-1]}")

directions.append(previous_segment.type.name)
segments = [previous_segment]
for i in range(8):
    
    next_segment_position = previous_segment.next_tile_map_position()
    next_segment_orientation = previous_segment.next_tile_orientation()

    tile_type = np.random.choice(list(TileType))

    next_segment = LaneSegment(
        next_segment_position, next_segment_orientation , tile_type)
    previous_segment = next_segment
    print(f"segment: {previous_segment.map_position}, {np.degrees(previous_segment.orientation)}, {previous_segment.type} start: {previous_segment.points[0]}, end: {previous_segment.points[-1]}")
    map.extend(previous_segment.points)
    directions.append(previous_segment.type.name)
    segments.append(previous_segment)


print(f'map was {directions}')
plt.scatter([p[0] for p in map], [p[1] for p in map])
```

```python
from LaneSegment import LaneSegment, TileType
from matplotlib import pyplot as plt
import cv2
import numpy as np  
np.random.seed(42)

def plot_segment_with_directions(points,color='blue'):
    points_with_direction = []
    for p in points:
        distance = 0.1
        theta = p[2]
        next_point = (p[0] + distance * np.cos(theta), p[1] + distance * np.sin(theta))
        points_with_direction.append(next_point)

    for i in range(len(points)):

        plt.arrow(points[i][0], points[i][1], 
                points_with_direction[i][0] - points[i][0], 
                points_with_direction[i][1] - points[i][1], color=color, head_width=0.01, head_length=0.03)


segment = LaneSegment(
    (0,0), 0 , TileType.Straight)
plot_segment_with_directions(segment.points,color='blue')
plt.scatter([p[0] for p in segment.points], [p[1] for p in segment.points])
segment = LaneSegment(
    (0,0), 0 , TileType.RightTurn)
```

```python
plot_segment_with_directions(segment.points,color='green')
plt.scatter([p[0] for p in segment.points], [p[1] for p in segment.points], color='green')
plt.scatter([segment.points[19][0]], [segment.points[19][1]], color='orange')
```

```python
segment = LaneSegment(
    (0,0), 0 , TileType.LeftTurn)
plot_segment_with_directions(segment.points,color='red')
plt.scatter([p[0] for p in segment.points], [p[1] for p in segment.points], color='red')
plt.show()
```
