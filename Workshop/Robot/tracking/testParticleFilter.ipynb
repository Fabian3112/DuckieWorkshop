```python
from ParticleFilter import ParticleFilter
from LaneSegment import LaneSegment
import numpy as np
np.random.seed(42)  # Set the random seed for reproducibility
pf = ParticleFilter(n_particles=500)
pf.reset(x=0.0, y=0.15, theta=0)
```

```python
from matplotlib import pyplot as plt
import cv2
import numpy as np

for i in range(12):
    pf.predict(distance=0.1, rotation=0.0)
    pf.calculate_weights()
    pf.normalize()
    current_segment, point =pf.estimate()
    print(f'Estimated segment: {current_segment.type}, position {current_segment.map_position}, orientation {current_segment.orientation}, \npoint: {point} \nstart: {current_segment.points[0]}, \nend: {current_segment.points[-1]}\n')
    pf.resample()
    images = pf.plot_particles()
    img = np.zeros((60, 60 * 3, 3), dtype=np.uint8)
    i = 0
    for key in images:
        print(f'key: {key}')
        img[:, i * 60:(i + 1) * 60, :] = images[key]
        i += 1

    plt.imshow(img)
    plt.show()



```

```python
def draw_map(pf):
    map, point = pf.estimate_full_map()

    points = []
    for key in map:
        tile = map[key]
        points.extend(tile.points)

    point = np.array(point)
    print(points)
    print(point)
    map_img = LaneSegment.draw_points(points,point)
    #cv2.circle(map_img,(int(point[0]*100), int(point[1]*100)), 5, (0, 0, 255), -1)

    return map_img
```

```python
plt.imshow(draw_map(pf))
```

```python
import time
import numpy as np
start = time.perf_counter()
distance = 0.1
rotation = np.radians(0.2)
num_iterations = 100

for i in range(num_iterations):
    pf.predict(distance, rotation)

t_predict = time.perf_counter()

for i in range(num_iterations):
    pf.calculate_weights()

t_weights = time.perf_counter()

for i in range(num_iterations):
    pf.normalize()

t_normalize = time.perf_counter()

for i in range(num_iterations):
    pf.estimate()

t_estimate = time.perf_counter()

for i in range(num_iterations):
    pf.resample()

t_resample = time.perf_counter()

#for i in range(num_iterations):
#    map,point = pf.estimate_full_map()

t_full_map = time.perf_counter()

#for i in range(num_iterations):
#    draw_map(pf)

t_full_map_draw = time.perf_counter()

print(
    f"predict:   {(t_predict-start)*1000:.1f} ms\n"
    f"weights:   {(t_weights-t_predict)*1000:.1f} ms\n"
    f"normalize: {(t_normalize-t_weights)*1000:.1f} ms\n"
    f"estimate:  {(t_estimate-t_normalize)*1000:.1f} ms\n"
    f"resample:  {(t_resample-t_estimate)*1000:.1f} ms\n"
    f"full map:  {(t_full_map-t_resample)*1000:.1f} ms\n"
    f"draw map:  {(t_full_map_draw-t_full_map)*1000:.1f} ms"

)
```
