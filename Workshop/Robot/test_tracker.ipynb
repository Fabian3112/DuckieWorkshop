```python
from tracker_2_test import ParticleFilter
import random
import numpy as np

np.random.seed(42)  # Set a seed for reproducibility
random.seed(42)  # Set a seed for reproducibility
pf = ParticleFilter(n_particles=100)

x = 0.0
y = 0.45
theta = 0.0

pf.reset(x, y, theta)
```

```python
import matplotlib.pyplot as plt
import cv2
pf.calculate_weights()
for i in range(12):
    pf.predict(distance=0.05, rotation=0.0)
    pf.calculate_weights()
    pf.normalize()
    images = pf.plot_particles()
    pf.resample()

    for key in images.keys():
        image = images[key]
        cv2.line(image, (int((i+1)/6*60), 0), (int((i+1)/6*60), 60), (255, 0, 0), 1)
        plt.imshow(image)
        plt.title(f"Particle Filter - {key}")
        plt.show()

    print(pf.estimate())
```

```python
import numpy as np
import cv2
def draw_map(pf):
    map, point = pf.estimate_full_map()

    min_x = min(tile.map_position[0] for tile in map.values())
    max_x = max(tile.map_position[0] for tile in map.values())
    min_y = min(tile.map_position[1] for tile in map.values())
    max_y = max(tile.map_position[1] for tile in map.values())

    map_img =  np.zeros(((max_y - min_y + 1) * 60, (max_x - min_x + 1) * 60, 3), dtype=np.uint8)

    for key in map:
        tile = map[key]
        img = tile.draw()
        x_offset = (tile.map_position[0] - min_x) * 60
        y_offset = (tile.map_position[1] - min_y) * 60
        map_img[y_offset:y_offset + 60, x_offset:x_offset + 60] = img

    cv2.circle(map_img,(int(point[0]*100), int(point[1]*100)), 5, (0, 0, 255), -1)

    return map_img
```

```python
img = draw_map(pf)
plt.imshow(img)
plt.show()
```

```python
for i in range(12):
    pf.predict(distance=0.06, rotation=-0.075)
    pf.calculate_weights()
    pf.normalize()
    pf.resample()
    images = pf.plot_particles()

    print(pf.estimate())


for key in images.keys():
    image = images[key]
    plt.imshow(image)
    plt.title(f"Particle Filter - {key}")
    plt.show()

img = draw_map(pf)
plt.imshow(img)
plt.show()
```

```python
for i in range(6):
    pf.predict(distance=0.1, rotation=0.0)
    pf.calculate_weights()
    pf.normalize()
    pf.resample()
    images = pf.plot_particles()

for key in images.keys():
    image = images[key]
    plt.imshow(image)
    plt.title(f"Particle Filter - {key}")
    plt.show()

    print(pf.estimate())

img = draw_map(pf)
plt.imshow(img)
plt.show()
```
