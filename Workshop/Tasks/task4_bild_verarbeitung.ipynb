#### Der HSV-Farbraum
Im RGB-Farbraum gibt es 3 Zahlen, die die Farbe eines Pixels beschreiben: Den Wert für Rot, den Wert für Grün un den Welt für Blau. So ist es im HSV-Farum auch. Es gibt 3 Zahlen, die die Farbe eines Pixels bestimmen. <br>
Zuerst kommt **Farbton** (H): Besagt, welche Farbe du siehst. Dieser wird in Grad gemessen (Werte von 0 bis 360; H=0: Farbe ist Rot). <br>
Dann **Sättigung** (S): Besagt, wie intensiv eine Farbe ist. (S=0: Farbe ist grau, S=100: Farbe ist sehr kräftig.) <br>
Dann **Helligkeit** (V): Beschreibt, wie hell oder dunkel eine Farbe ist. (V=0: Farbe ist schwarz, V=100: Farbe ist weiß)

```python
hsv = cv.imread("images/hsv.png")
hsv = cv.cvtColor(hsv, cv.COLOR_BGR2RGB)
plt.imshow(hsv)
```

### Jetzt nutzen wir Python-Wissen und die Bibliothek OpenCV um einen Spur-Finde-Algorithmus zu entwerfen!

Du bekommst als **Eingabe ein Bild**, auf welchem eine Spur zu sehen ist.

Dein **Ziel** ist es in diesem Bild einen Punkt zu finden, welcher der Mittelpunkt der Spur ist. Also ein Punkt, zu welchem der Duckiebot hinfahren soll! Wenn die Spur gerade aus geht, dann soll der Punkt auch ziemlich in der Mitte des Bildes liegen; Wenn die Spur allerdings eine Kurve macht, dann sollte der Punkt etwas weiter links, bzw rechts liegen. 

```python
# Beispiel fur eine gerade Spur
gerade = cv.imread("images/gerade.jpeg")
gerade = cv.cvtColor(gerade, cv.COLOR_BGR2RGB)
plt.imshow(gerade)
```

```python
# Beispiel für eine Kurve
bisschen_kurve = cv.imread("images/bisschen_kurve.jpeg")
bisschen_kurve = cv.cvtColor(bisschen_kurve, cv.COLOR_BGR2RGB)
plt.imshow(bisschen_kurve)
```

Zuerst musst du die weiße und die gelbe Spur unterscheiden können. Nutze hierzu den HSV-Farbraum, den du oben kennen gelernt hast. 

Für welche Werte von H, S und V kann man gelb sehen? Welche Werte bilden weiß? (Nutze hierzu gerne google, oder eine andere Suchmaschine deiner Wahl, um passende Startwerte zu finden).

Mit der Funktion make_mask_image könnt ihr eure hsv Werte testen: zeigt die Maske den richtigen Streifen? Dann habt ihr gute Werte gefunden.

```python
# Hier ist eine Funktion, die ein Bild und HSV-Grenzen als Eingabe nimmt und eine Maske zurückgibt, die alle Pixel einschließt, die innerhalb der angegebenen HSV-Grenzen liegen.
def make_mask_image(img, h_min, s_min, v_min, h_max, s_max, v_max):
    # Konvertiere das Bild von RGB nach HSV
    img_hsv = cv.cvtColor(img, cv.COLOR_RGB2HSV)
    
    # Erstelle eine Maske, die alle Pixel einschließt, die innerhalb der angegebenen HSV-Grenzen liegen
    lower_bound = (h_min, s_min, v_min)
    upper_bound = (h_max, s_max, v_max)
    mask = cv.inRange(img_hsv, lower_bound, upper_bound)
    
    return mask
```

```python
# Hier gibt es Platz um auszuprobieren was mask tut und wie ihr am besten herausfindet welche HSV-Grenzen ihr braucht um die gewünschten Pixel zu maskieren.
```
