## Kreuzungen navigieren

In dieser Aufgabe sollt ihr den Roboter dazu bringen an kreuzungen korrekt abzubiegen. Wie in den vorherigen Aufgaben verwenden wir ein Roboter objekt. Der Roboter kann bereits gerade aus fahern (robot.forward) sich drehen (robot.left / robot.right) und der Straße folgen (robot.start() / robot.stop())

```python
from Robot.robot import Robot
import time

# ein roboter object erstellen
robot = Robot()

# der roboter kann wie die Zeichnen-Schildkröte forwärts fahren und sich drehen
robot.forward(1)
robot.left(90)
robot.right(90)
```

Die Methode wait_for_redline läuft so lange, bis der Roboter eine rote Linie sieht

```python
def wait_for_redline(robot:Robot):
    redline = robot.get_redline()

    while not redline:
        time.sleep(0.1)
        redline = robot.get_redline()

    return redline

print("Waiting for redline...")
wait_for_redline(robot)
print("Redline detected!")
```

Die Methode get_intersection des Roboters gibt ein array mit allen Abbiege möglichkeiten zurück. Ihr könnt die Methode np.random.choice() verwenden um aus den möglichkeiten eine auszuwählen. Die Abbiege möglichkeiten sind: **left** **right** **straight** 

```python
import numpy as np

wait_for_redline(robot)
intersection = robot.get_intersection()
turn = np.random.choice(intersection)

print(f"Intersection detected! Turning {turn}...")
```

Jetzt seid ihr dran. Lasst den Roboter 3 Runden fahren wobei er an Roten linien 3 Sekunden warten soll und an den kreuzungen korrekt abbiegen soll.

```python
#TODO remove solution
for i in range(10):
    robot.start()
    wait_for_redline(robot)

    robot.stop()
    print("Intersection detected! Waiting 3 seconds before turning...")
    time.sleep(3)
    intersection = robot.get_intersection()
    turn = np.random.choice(intersection)
    print(f"Turning {turn}...")
    
    if turn == "left":
        robot.forward(0.4)
        robot.left(90)
        robot.forward(0.2)

    elif turn == "right":
        robot.forward(0.2)
        robot.right(90)

    elif turn == "straight":
        robot.forward(0.4)
```
