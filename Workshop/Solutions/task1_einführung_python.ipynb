#### Schleifen
Eine Schleife ist ein Befehl der dem Computer sagt, dass er etwas wiederholen soll. Er wiederholt es so oft, bis du ihm sagst, dass er aufhören soll. Statt also etwas oft hintereinander zu schreiben, kannst du es in eine Schleife schreiben. <br>
Beispiel: Du willst allen Schülern einmal Hallo sagen oder zählen wieviele Schüler es gibt.

Wie du gesehen hast, können Variablen in Schleifen genutzt werden.

```python
for i in range(anzahl):
    print("hello")

anzzahl_gezahlt = 0
for i in range(anzahl):
    anzzahl_gezahlt = anzzahl_gezahlt + 1
print(anzzahl_gezahlt)
```

#### Methoden
Eine Methode ist eine Anleitung, die man einmal hinschreibt und dann immer wieder verwenden kann. Stell es dir vor wie ein Rezept in der Küche, wenn du ein Rezept für Pfannkuchen hast, musst du dir nicht jedes mal neu überlegen wie es geht - du folgst einfach dem Rezept! 

```python
def sage_hallo(anzahl):
    for i in range(anzahl):
        print("hello")
```

```python
# Diese Zeile ruft die Methode auf. Sie backt dann tatsächlich den Kuchen. ;)
sage_hallo(5)
```

```python
sage_hallo(anzahl=anzzahl_gezahlt)
```

#### Array
Ein Array ist wie eine Liste oder Reihe von Dingen, die zusammengehören. Stell dir vor du hast eine Reihe von Obstsorten: 🍎 🍌 🍇 🍊. Diese kannst du nun in ein Array legen:

```python
obst = ["Apfel", "Banane", "Traube", "Orange"]
```

Jetzt weiß der Computer in dem Array obst liegen 4 Dinge. <br>
Du kannst dir das Array als Schublade mit vielen Fächern vorstellen. Jedes Fach hat dabei eine Nummer und du kannst sagen: Gib mir was sich in Fach 2 befindet: <br>
Bemerke hierbei: Angefangen zu zählen wird hier bei 0.

```python
print(obst[0])
print(obst[1])
print(obst[2])
print(obst[3])
```

Jetzt kannst du aber auch größere Teile eines Arrays anschauen: <br>
Mit dem Doppelpunkt (:) kannst du angeben von welcher Anfangszahl bis zu welcher Endzahl du in die Schubladen schauen willst. 

```python
print(obst[1:3])
```

```python
print(obst[:2])
```

#### if-Bedingung
Eine if-Bedingung ist wie eine Entscheidung. Du gibst dem Computer also nur eine Frage und er entscheidet, was er tun soll, je nachdem ob die Antwort "ja" oder "nein" ist. Stell dir vor du stehst an einer Ampel: Wenn die Ampel grün ist, gehst du, wenn die Ampel rot ist, bleibst du stehen.

```python
ampel = "grün"

if ampel == "grün":
    print("Geh!")
else:
    print("Bleib stehen!")
```

Weiteres Beispiel: Ist die Zahl größer als 10?

```python
zahl = 12

if zahl > 10:
    print("Die Zahl ist größer als 10!")
else:
    print("Die Zahl ist 10 oder kleiner!")
```

Wenn du mehr als 2 Möglichkeiten hast, sieht das ganze dann so aus:

```python
zahl = 10

if zahl > 10:
    print("Die Zahl ist größer als 10!")
elif zahl == 10:
    print("Die Zahl ist 10!")
else:
    print("Die Zahl ist kleiner als 10!")
```
