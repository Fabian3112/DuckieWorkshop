Euer Duckiebot verfügt über eine kamera und zwei Motoren. Mithilfe der Kamera kann der Duckiebot seine Fahrspur erkennen und ihr folgen. Euer Roboter kann bereits einige Aufgaben. 

Um mit dem Roboter zu arbeiten erstellt ihr zunächst ein Roboter object wie bei der Zeichnen-Schildkröte. Ähnlich wie die Zeichnen-Schildkröte unterstützt kann der Roboter bereits einigige funktionen, mit denen ihr euch nun vertraut machen könnt.

```python
from Robot.robot import Robot

# ein roboter object erstellen
robot = Robot()

# der roboter kann wie die Zeichnen-Schildkröte forwärts fahren und sich drehen
robot.forward(1)
robot.left(90)
robot.right(90)
```

Wir können uns auch das Kamera bild des Roboters anschauen

Euer Roboter kann bereits der Straße folgen. Um den Roboter losfahren zu lassen könnt ihr die Methode start() verwenden. Mit stop() hält der Roboter an

```python
import time

robot.start()
time.sleep(20)
robot.stop()
```

Der Roboter kann außerdem bereits Rote linien erkennen. Die methode wait_for_redline läuft so lange, bis der Roboter eine Rote linie sieht. Testet die methode in dem ihr den Roboter zunächst etwas entfernt der Roten Linie auf die Spur stellt und bewegt ihn dann zur Roten Linie bis ihr die Ausgabe **Redline detected!** seht

```python
def wait_for_redline(robot:Robot):
    redline = robot.get_redline()

    while not redline:
        time.sleep(0.1)
        redline = robot.get_redline()

    return redline

print("Waiting for redline...")
wait_for_redline(robot)
print("Redline detected!")
```
